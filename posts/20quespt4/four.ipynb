## Reflection 

Okay… I’ll admit it: this round was almost *embarrassingly* smooth. Like, compared to my past 20Qs where the LLm didnt always ask all of their alotted questions, questions didnt build off each other, etc, this felt like watching two robots politely play chess. Gemini came in with the classic “start broad, then narrow down” strategy and just… didn’t miss. Living thing? Man-made? Touchable? Indoors? Bigger than a shoebox? It’s giving *efficient*, it’s giving *no wasted moves*, it’s giving “I actually read the rules.” And Grok, to its credit, stayed consistent and didn’t do any weird loophole stuff like “well technically it’s wood AND fabric”

The result: Gemini correctly landed on “sectional sofa” by Question 18, which is honestly pretty impressive because “sectional” is a specific subtype, not just “couch.” 

## Why this might have worked *so* well 

Also… I can’t ignore the obvious: we used two popular commercial models. These models tend to be **bigger**, more heavily trained, and more aggressively fine tuned for “clean conversation” than a lot of the smaller or more experimental models I’ve played with in earlier posts. Bigger models usually mean better performance on stuff like: keeping track of constraints, staying consistent across many turns, and choosing high-information questions that slice the search space quickly. So part of the success here might be less “wow, LLMs are suddenly amazing,” and more “wow, we picked the two models most likely to behave like responsible adults in a structured task.”
