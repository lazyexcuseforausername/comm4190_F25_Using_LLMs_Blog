## Introduction

What happens when you take the guardrails off a language model and let it answer moral questions without censoring itself? On the surface, an “uncensored” LLM sounds appealing—more honest, more direct, less filtered by corporate safety rules or polite neutrality. Today we will be exploring this with the uncensored llama 3.1 model.
