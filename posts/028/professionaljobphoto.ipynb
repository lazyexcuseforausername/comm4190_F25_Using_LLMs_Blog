## Introduction

Okay so for this blog I wanted to do something a little different (and honestly more useful for real life): a fun experiment with **ChatGPT 5.2 Thinking** where I see how much it can help with the “professional job search” things that I always end up procrastinating (specifically, getting a solid headshot and prepping for interviews.)

The purpose of this post is basically to test if ChatGPT can act like a lowkey career helper. Like, can it help me figure out what makes a headshot look professional without it looking stiff or giving “corporate stock photo”?Can it help with the practical stuff too like lighting, background, posing, what to avoid, and what actually matters versus what people just overthink?

Then there’s the interview side, which I think is even more interesting. I want to see if ChatGPT can help me get an idea of what kinds of interview questions I should be practicing for the specific career that I want (and provide example answers so I can gauge where I am in my career journey and where I need to be.)
